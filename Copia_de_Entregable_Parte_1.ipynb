### Conteo de frecuencias de los lenguajes de programación

La columna que contiene información sobre los lenguajes de programación utilizados es `tools_programming_languages`. Sus valores son strings con los lenguajes seleccionados separados por comas.



```
df.tools_programming_languages[:3]
```

Las siguientes celdas de código separan estos lenguajes de programación y cuentan la frecuencia con la que aparecen.

No es necesario entender este código en profundidad, aunque sí es un buen ejercicio.

```
# Convert the comma-separated string of languages to a list of string.
# Remove 'ninguno de los anteriores' option, spaces and training commas.
def split_languages(languages_str):
  if not isinstance(languages_str, str):
    return []
  # Remove 'other' option
  languages_str = languages_str.lower()\
    .replace('ninguno de los anteriores', '')
  # Split string into list of items
  # Remove spaces and commas for each item
  return [lang.strip().replace(',', '')
          for lang in languages_str.split()]

# Create a new column with the list of languages
df.loc[:, 'cured_programming_languages'] = df.tools_programming_languages\
    .apply(split_languages)
if 'cured_programming_languages' not in relevant_columns:
    relevant_columns.append('cured_programming_languages')

# Duplicate each row of df for each programming language
# mentioned in the response.
# We only include in df_lang the columns we are going to analyze later, so we
# don't duplicate innecesary information.
df_lang = df.cured_programming_languages\
    .apply(pd.Series).stack()\
    .reset_index(level=-1, drop=True).to_frame()\
    .join(df[relevant_columns])\
    .rename(columns={0: 'programming_language'})
# Horrible programming style! But a lot of data science code can be written with
# as concatenations of functions (pipelines), and there's no elegant way of
# doing that on Python.
df_lang[:5]
```

En la columna `programming_language` se encuentra cada lenguaje por separado. Notar que si una respuesta contenía 3 lenguajes, como `"HTML, Javascript, Python"`, la fila ha sido replicada 3 veces. Por ello, hay tres filas con índice 1.

```
language_count = df_lang.programming_language.value_counts()\
    .reset_index()\
    .rename(columns={'index': 'language', 'programming_language': 'frequency'})
language_count[:10]
```

- Ahora realizaremos algunas suposiciones sobre los datos, como así también algún filtrado de datos NaN.

```
# Filtrado de NaN's
df_data = df_data.dropna(subset=['programming_languages', 'salary_monthly'])
df_data = df_data.reset_index(drop=True)
```

```
# Filtrado de sueldos que no tengan sentido, a continuación vemos que existen outliers importantes
df_data['salary_monthly'].describe()
```

- Para eliminar los outlier utilizaremos el [método de Tukey](https://www.linkedin.com/pulse/what-tukey-outlier-jason-shafrin).

```
q1 = df_data['salary_monthly'].quantile(0.25)
q3 = df_data['salary_monthly'].quantile(0.75)
iqr = q3 - q1

# Calculamos los límites mediante el coeficiente de sensibilidad default
lower_limit = q1 - 1.5 * iqr
upper_limit = q3 + 1.5 * iqr

print(lower_limit, upper_limit)
```

- El límite inferior del salario da negativo, esto es por que los datos se concentran la parte inferior de la distribución. Por lo tanto, considearemos que un salario mínimo como $100k.

```
lower_limit = 100000

# Intentamos filtrar los datos Outliers
df_data = df_data[(df_data['salary_monthly'] >= lower_limit) & (df_data['salary_monthly'] <= upper_limit)]
df_data = df_data.reset_index(drop=True)
```

```
sns.set(style="whitegrid")

stats = df_data['salary_monthly'].describe().round(2).to_string()

fig = plt.figure(figsize=(10, 6))
gs = fig.add_gridspec(nrows=2, ncols=2, width_ratios=[4, 1], height_ratios=[1, 3])
ax_box = fig.add_subplot(gs[0, 0])
ax_hist = fig.add_subplot(gs[1, 0])
ax_text = fig.add_subplot(gs[:, 1])

# Boxplot
sns.boxplot(x=df_data['salary_monthly'], ax=ax_box, color='skyblue')
ax_box.set(title='Distribución y Boxplot del Salario mensual')
ax_box.set(xlabel='')
ax_box.tick_params(axis='x', which='both', bottom=False, top=False, labelbottom=False)

# Histograma
bins = int(len(df_data) / 100)
ax_hist.hist(df_data['salary_monthly'], bins=bins, color='skyblue', edgecolor='black')
ax_hist.set(xlabel='Salario Mensual', ylabel='Frecuencia')

# Stats
ax_text.axis('off')
ax_text.text(0, 1, "Estadísticos\n\n" + stats, fontsize=10, va='top')

plt.tight_layout()
plt.show()
```

- En muchos trabajos, el bono es un aporte bastante importante, de aquí ya empezaremos a realizar suposiciones "ingenieriles".

```
# Realizaremos algunas suposiciones sobre cuanto aportan los bonos medidos en cantidad de salarios netos mensuales
bonus_map = {
    'No': 0.0,
    'Menos de un sueldo': 0.5,
    'Un sueldo': 1.0,
    'De uno a tres sueldos': 2.0,
    '3+ sueldos': 3.0
}

df_data['bonus'] = df_data['bonus'].replace(bonus_map).astype(float)
```

- Generalmente los cálculos de salarios se realizan en base a las horas trabajadas, por lo cual se tendrá en cuenta la dedicación para calcular cual es el valor del trabajo medido en horas. A partir de datos obtenidos obtenidos de la siguiente [fuente](https://gusto.com/es/resources/articles/hr/team-management/difference-part-time-full-time-employee), podemos aproximar que el trabajo full-time corresponde a 130 horas al mes. El trabajo part-time se define como aquel que dure menos que esas horas, podemos realizar una suposición de que en promedio podría ser de un 25% menos.

```
# Realizaremos algunas suposiciones sobre la dedicación, si es full-time corresponden 130hs, si es part-time aproximamos a un 20% menos
dedication_map = {
    'Full-Time':130.0,
    'Part-Time':130*0.75,
}

df_data['dedication'] = df_data['dedication'].replace(dedication_map).astype(float)
```

- Se generará una columna que contenga el salario medido por horas, y que además tiene en cuenta la ganancia anual en forma de bonos. Estos bonos son fracciones de sueldos mensuales.

$$
\text{Salario por hora} = \frac{\text{salario mensual} \times (12 + \text{bonus})}{12 \times \text{dedication}}
$$


```
df_data['salary_per_hour'] = ((df_data['salary_monthly'] * (12 + df_data['bonus'])) / (12 * df_data['dedication'])).round().astype(int)
```

```
sns.set(style="whitegrid")

stats = df_data['salary_per_hour'].describe().round(2).to_string()

fig = plt.figure(figsize=(10, 6))
gs = fig.add_gridspec(nrows=2, ncols=2, width_ratios=[4, 1], height_ratios=[1, 3])
ax_box = fig.add_subplot(gs[0, 0])
ax_hist = fig.add_subplot(gs[1, 0])
ax_text = fig.add_subplot(gs[:, 1])

# Boxplot
sns.boxplot(x=df_data['salary_per_hour'], ax=ax_box, color='lightgreen')
ax_box.set(title='Distribución y Boxplot del Salario por Hora')
ax_box.set(xlabel='')
ax_box.tick_params(axis='x', which='both', bottom=False, top=False, labelbottom=False)

# Histograma
bins = int(len(df_data) / 100)
ax_hist.hist(df_data['salary_per_hour'], bins=bins, color='lightgreen', edgecolor='black')
ax_hist.set(xlabel='Salario por Hora', ylabel='Frecuencia')

# Stats
ax_text.axis('off')
ax_text.text(0, 1, "Estadísticos\n\n" + stats, fontsize=10, va='top')

plt.tight_layout()
plt.show()
```

```
import plotly.graph_objects as go
import plotly.express as px

# Explode y agrupar
df_exploded = df_data.explode("programming_languages")
grouped = df_exploded.groupby("programming_languages").agg(
    popularity=("programming_languages", "count"),
    avg_salary=("salary_per_hour", "mean")
).reset_index()

# Encontrar los destacados
most_popular = grouped.loc[grouped["popularity"].idxmax()]
highest_paid = grouped.loc[grouped["avg_salary"].idxmax()]

# Scatter plot
scatter = px.scatter(
    grouped,
    x="avg_salary",
    y="popularity",
    hover_name="programming_languages",
    labels={
        "avg_salary": "Salario promedio por hora",
        "popularity": "Popularidad"
    },
    title="Popularidad vs Salario",
    log_y=True,
    width=1000,
    height=700
)

scatter.update_traces(marker=dict(size=10, opacity=0.7, color='mediumseagreen'))
scatter.update_layout(showlegend=False)

# Crear figura con subplots para incluir la tabla
fig = go.Figure(data=scatter.data)

# Agregar la tabla como anotación (sin flechas)
table_data = [
    ["Tipo", "Lenguaje", "Popularidad", "Salario promedio"],
    ["Más popular", most_popular["programming_languages"], most_popular["popularity"], f"${most_popular['avg_salary']:.2f}"],
    ["Mejor pago", highest_paid["programming_languages"], highest_paid["popularity"], f"${highest_paid['avg_salary']:.2f}"]
]

fig.add_trace(
    go.Table(
        header=dict(values=table_data[0], fill_color="paleturquoise", align="left"),
        cells=dict(values=list(zip(*table_data[1:])), fill_color="lavender", align="left"),
        domain=dict(x=[0.7, 1.0], y=[0.3, 0.9])  # Posición en el gráfico
    )
)

fig.show()
```
