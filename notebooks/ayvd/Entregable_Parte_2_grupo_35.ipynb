## Ejercicio 1: Estimación

**Consigna:**  Calcular una estimación puntual y un intervalo de confianza de nivel (1-alpha) para la resta entre la media del salario Neto para Hombres y la media del salario Neto para mujeres (diferencia de las medias entre el grupoA y grupoB).
¿Cómo se relaciona este intervalo de confianza con el test de hipótesis para diferencia de medias?

# Respuesta - Ejercicio 1

```
import numpy as np
from scipy import stats
import pandas as pd

# Cambiamos nombres de las variables para que sea más claro
men_salary = groupA
woman_salary = groupB
```

Calcularemos algunos estadísticos sobre los grupos.

```
# Media
x_m = np.mean(men_salary)
x_w = np.mean(woman_salary)

# Desviación estándar
s_m = np.std(men_salary, ddof=1)
s_w = np.std(woman_salary, ddof=1)

# Tamaño de las muestras
n_m = len(men_salary)
n_w = len(woman_salary)
```

Realizamos la estimación puntual de la diferencia de medias.


```
diff_medias = x_m - x_w
print(f"Estimación puntual de la diferencia de medias (Hombres - Mujeres): {diff_medias:.2f}")
```

Intervalo de confianza utilizando t de Welch, tal como se hizo en el práctico.

```
# Error estándar de la diferencia
se_diff = np.sqrt((s_m**2 / n_m) + (s_w**2 / n_w))

# Grados de libertad (aproximación de Welch)
df = ((s_m**2 / n_m + s_w**2 / n_w)**2) / (((s_m**2 / n_m)**2 / (n_m - 1)) + ((s_w**2 / n_w)**2 / (n_w - 1)))

# Valor crítico para el intervalo de confianza
t_crit = stats.t.ppf(1 - alpha / 2, df)
```

Calculamos el margen de error.

```
# Margen de error
margen_error = t_crit * se_diff
```

Calculamos el intervalo de confianza.

```
# Intervalo de confianza
ic_inferior = diff_medias - margen_error
ic_superior = diff_medias + margen_error
print(f"Intervalo de confianza al {100*(1-alpha)}%: ({ic_inferior:.2f}, {ic_superior:.2f})")
```

Como resultado tenemos un intervalo de confianza de 95%. De 100 muestras aleatorias distintas, 95 tendrían en su IC el valor real de la población que nos interesa.

En respuesta a la consigna “¿Cómo se relaciona este intervalo de confianza con el test de hipótesis para diferencia de medias?”, planteamos los siguientes puntos:

En este análisis estamos evaluando la diferencia de salarios, utilizando la media como medida representativa / estadístico.

Estos salarios tienen distribuciones que pueden o no tener medias iguales.

Por lo tanto, es razonable considerar **una hipótesis nula que plantea que la diferencia de medias entre ambas distribuciones es igual a cero**, es decir, que las medias son iguales.

Como observamos que el valor cero no se encuentra dentro del intervalo de confianza obtenido, podemos concluir que se rechaza la hipótesis nula de igualdad de medias con un nivel de significación del 5%.

$H_0 =$ Las medias de los salarios entre hombres y mujeres son iguales. ❌\
$H_1 =$ Las medias de los salarios entre hombres y mujeres difieren. ✅

# Respuesta - Ejercicio 2

1. Definimos $H_0, H_1$

- **Hipótesis Nula:** la distribución de salarios entre los hombres y mujeres son iguales,

$$H_0 : \mu A = \mu B$$

- **Hipótesis Alternativa:**  la media de los salarios entre hombres y mujeres son diferentes,

$$H_1 : \mu A \neq \mu B$$

***

2. Usamos el estadístico $t$ de Welch, ya que las varianzas de las poblaciones pueden diferir:

$$t = \frac{\bar{X}_A - \bar{X}_B}{\sqrt{\frac{s_A^2}{n_A} + \frac{s_B^2}{n_B}}}$$

```
# Calcular test t de Welch                                  False => varianzas distintas
t_stat, p_value = stats.ttest_ind(men_salary, woman_salary, equal_var=False)

# Mostrar resultados
print(f"Estadístico t: {t_stat:.4f}\n")
print(f"P-valor: {p_value:.6f}\n")

if p_value < alpha:
    print("Rechazamos la hipótesis nula: hay diferencias significativas en los salarios.\n")
else:
    print("No se puede rechazar la hipótesis nula: no hay evidencia suficiente de diferencias.\n")
```

**Conclusión:** El p-valor es extremadamente pequeño, esto quiere decir que la probabilidad de que obtener una diferencia de medias tan grande, si H0 fuera cierta, es extremadamente baja. En principio nos acotamos a decir que p-valor es menor al nivel de significación del 5%.
